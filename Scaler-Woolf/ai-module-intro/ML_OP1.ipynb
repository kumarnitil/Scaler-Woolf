# MLOps and Machine Learning Overview

## **Definition of MLOps**
*   **MLOps** is defined as the combination of **Machine Learning (ML)** and **Operations (Ops)**.

## **The MLOps Lifecycle**
The workflow for processing data and deploying models follows a specific sequence:
*   **Data Input:** The process starts with **trained data**, which typically contains **noise** such as duplicate records, null values, or special characters.
*   **Data Cleaning:** It is essential to **clean the noisy data** before moving to the next stage.
*   **Modeling:** Once the data is prepared, an **AI Model (ML)** is developed using various **ML Algorithms**.
*   **Deployment:** The final stages involve **saving the model** and then moving to **deploy production**.

## **Evolution of AI Technologies**
The field has progressed through several stages of complexity:
*   **Machine Learning (ML)** $\rightarrow$ **Deep Learning (DL)** $\rightarrow$ **Generative AI (Gen AI)** $\rightarrow$ **Agentic AI**.

## **Core Machine Learning Categories and Algorithms**
Machine Learning is structured around different types of algorithms used to build AI models:

### **1. Classification**
*   Used for categorizing data into specific groups.
*   **Examples:** KNN, Decision tree, and Naïve Bayes.

### **2. Regression**
*   Used for predicting continuous values.
*   **Examples:** Linear regression and logistic regression.

### **3. Clustering**
*   Used for grouping similar data points together.
*   **Example:** K-means.

### **4. Association**
*   Used for discovering rules that describe your data.
*   **Examples:** association_rules and Multinomial.

## Data Cleaning and Manipulation (Practical Example)

*   **Pandas Library:** Used to **clean, manipulate, and organize data** into a **tabular format** by converting raw data structures (like dictionaries) into a **DataFrame**.
*   **Noisy Data Identification:** Before developing an AI model, it is essential to **clean the noisy data** found within the dataset.
*   **Common Data Issues to Address:**
    *   **Extra Spaces:** Leading or trailing whitespace in strings, such as `"   Jyoti"`.
    *   **Missing or Null Values:** Data points represented as **`None`**, empty entries, or placeholder strings like **`","`**.
    *   **Duplicate Records:** Identical entries for the same entity, such as repeating an employee's name, department, and salary.
    *   **Special Characters:** Unwanted symbols like that may appear in trained datasets and require removal.
